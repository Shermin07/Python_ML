### 1. What is a Pandas Series?
A Pandas Series is a one-dimensional labeled array in Python that can store data of different types, such as integers, strings, and floats.

### 2. What is a Pandas DataFrame?
A Pandas DataFrame is a two-dimensional labeled data structure in Python that stores data in rows and columns, similar to an Excel spreadsheet or a table

### 6. Which method is used to iterate over rows in a DataFrame?

Correct answer: iterrows()

import pandas as pd

data = {
    'Name': ['John', 'Alice', 'David'],
    'Age': [20, 25, 30]
}

df = pd.DataFrame(data)

for index, row in df.iterrows():
    print(index, row['Name'], row['Age'])

4. Using .loc[] and .iloc[], show the first row, the second column, and a 2×2 block    from the middle of your DataFrame

```python
import pandas as pd

data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}

df = pd.DataFrame(data)

# First row using loc
print("First Row using loc:")
print(df.loc[0])

# First row using iloc
print("\nFirst Row using iloc:")
print(df.iloc[0])

# Second column using loc
print("\nSecond Column using loc:")
print(df.loc[:, 'Price'])

# Second column using iloc
print("\nSecond Column using iloc:")
print(df.iloc[:, 1])

# 2x2 block from the middle using loc
print("\n2x2 Block using loc:")
print(df.loc[1:2, 'Price':'Quantity'])

# 2x2 block from the middle using iloc
print("\n2x2 Block using iloc:")
print(df.iloc[1:3, 1:3])
```


5. Create a 4×3 DataFrame of student names and marks. Drop one column using .drop(), drop one row by index

```python
import pandas as pd

data = {
    'Name': ['John', 'Alice', 'David', 'Sara'],
    'Maths': [80, 90, 75, 85],
    'Science': [85, 95, 70, 80]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Drop one column
df1 = df.drop('Maths', axis=1)

print("\nAfter dropping Maths column:")
print(df1)

# Drop one row by index
df2 = df.drop(1, axis=0)

print("\nAfter dropping row with index 1:")
print(df2)
```


6. Create a DataFrame of employees with columns "Name", "Department", and "Salary". Sort it by "Salary" in ascending order and by "Name" in descending order

```python
import pandas as pd

data = {
    'Name': ['John', 'Alice', 'David', 'Sara'],
    'Department': ['IT', 'HR', 'Sales', 'IT'],
    'Salary': [50000, 40000, 60000, 45000]
}

df = pd.DataFrame(data)

# Sort by Salary in ascending order
salary_sort = df.sort_values(by='Salary', ascending=True)

print("Salary in Ascending Order:")
print(salary_sort)

# Sort by Name in descending order
name_sort = df.sort_values(by='Name', ascending=False)

print("\nName in Descending Order:")
print(name_sort)
```

Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.

```python
import pandas as pd

data = {
    'Name': ['John', 'Alice', 'David', 'Sara',
             'Mike', 'Emma', 'Peter', 'Sophia'],
    'Score': [85, 35, 70, 25, 95, 55, 30, 75]
}

df = pd.DataFrame(data)

# Students with Score > 60
print("Students with Score greater than 60:")
print(df[df['Score'] > 60])

# Students with Score < 40
print("\nStudents with Score less than 40:")
print(df[df['Score'] < 40])
```
