### M-9.4 OOP Inheritance..pillers(Inheritance, polymorphism, abstraction, encapsulation)

```python
### child class/derived class::here, SmartPhone = child class and Phone = parent class so that, SmartPhone can inherite all capabilities. 

class SmartPhone(Mobile):
    def __init__(self,model,bettery,camera,processor):
        super().__init__(model,bettery,camera)
        self.processor = processor

    def charge(self, hour):
        print("fast charging in process")
        super().charge(hour)

pro = SmartPhone("X", 5000, 25,"snapDragon")
print(pro.model)
pro.charge(1)

```

```python
### Single inheritance::


class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):
    def __init__(self, name, age, roll, marks):
        super().__init__(name, age)
        self.roll = roll
        self.marks = marks

    def show_result(self):
        print("Roll:", self.roll)
        print("Marks:", self.marks)


s1 = Student("Rahim", 18, 101, 85)

s1.introduce()
s1.show_result()
```

```python
### Multiple inheritance


class Phone:
    def __init__(self, brand):
        self.brand = brand

    def call(self, number):
        print(self.brand, "calling", number)


class Camera:
    def __init__(self, resolution):
        self.resolution = resolution

    def take_photo(self):
        print("Photo taken at", self.resolution, "MP")


class SmartPhone(Phone, Camera):
    def __init__(self, brand, resolution, model):
        Phone.__init__(self, brand)
        Camera.__init__(self, resolution)
        self.model = model

    def show_info(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Camera:", self.resolution, "MP")


p1 = SmartPhone("Samsung", 108, "Galaxy S")

p1.show_info()
p1.call("123456789")
p1.take_photo()

```

### M--9.5 OOP Polymorphism

## it means different form

```python
class Camera():
    def __init__(self,name):
        self.name = name

### method:
    def capture(self):
        print("A photo is captured")
        
```

```python
class Smart_Phone(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution

    def capture(self):
        print("A photo is captured by smart phone")


class Dslr(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution
    def capture(self):
            print("A photo is captured by dslr")

class Drone(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution
     
    def capture(self):
            print("A photo is captured by drone")

phone = Smart_Phone("phone", 30)
dslr = Dslr("dslr",40)
drone = Drone("drone",50)

phone.capture()
dslr.capture()
drone.capture()

```

### (Optional) 9.7 OOP Abstraction

```python
from abc import ABC, abstractmethod

class Telephone(ABC):
    @abstractmethod
    def make_call(self):
        pass

class Sphone(Telephone):
    def make_call(self):
        print("making a call using Sphone")


class Iphone(Telephone):
    def make_call(self):
        print("making a call using Iphone")

ips = Sphone()
ips.make_call()
ip = Iphone()
ip.make_call()

        
```
