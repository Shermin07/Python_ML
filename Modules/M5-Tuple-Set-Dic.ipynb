## M--5.4 Dictionary Methods

```python
## if has no values:
dic = {"course_name" : "phitron", "duration" : 6 , "subject" : "AI/ML", "module" : [1,2,3,4,5] }
print(dic.get("age"))
print(dic.get("age",0))
```

```python
## adding values:

dic["class_time"] = "everyday"
print(dic)

dic.update({"exams" : "thursday"})
print(dic)
```

```python
## deleting

del dic["class_time"]
print(dic)
```

```python
## copying

dic_2 = dic.copy()
print(dic_2)
```

```python
## testing::  value can be list, set, tuple and key can be only tuple because its immutable

dic_3 = {[1,2,3]: "a"}
print(dic_3)

```

```python
dic_4 = {(11,22,33): "b"}
print(dic_4)
```

## M--5.5 Dictionary View Methods

## iterating over dictionaries::





```python
dic = {"course_name" : "phitron", "duration" : 6 , "subject" : "AI/ML", "module" : [1,2,3,4,5] }

## keys:
keys = dic.keys()
print(keys)


dic["stating_month"] = 10
print(keys)

## values:
values = dic.values()
print(values)

## items

items = dic.items()
print(items)
```

```python
## iteration:
dic = {"course_name" : "phitron", "duration" : 6 , "subject" : "AI/ML", "module" : [1,2,3,4,5] }
for key,value in dic.items():
    print(key, value)
```

## M--5.6 Dictionary Comprehension


## iteration:



```python
print(list(range(10)))
print(range(10))

square = { x:x**2 for x in range(10)}
print(square)

square1 = { x:x**2 for x in range(1,11) if x%2 == 0}
print(square1)
```

```python
co_ordinates = [(1,2),(3,4)]
locations = {"Dhaka", "Finland"}

exact_location = {co_or:loc for co_or,loc in zip(co_ordinates,locations)}
print(exact_location)
```

### M--5.7 Problem solving with In Built Structures

```python
## given a list, make it unique list
list1 = [1, 2, 3, 2, 4, 3]

unique_lst = set(list1)
print(unique_lst)

final_list = list(unique_lst)
print(final_list)
```

```python
## given a string, print the frequency of the words:

string = """data science mechine learning data analysis data structure  mechine learning 
statistics data models data validation feature feature """


words = string.split()
print(words)

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```
