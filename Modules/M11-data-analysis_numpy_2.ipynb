```python
### 4. Create a 4×4 matrix and extract elements

import numpy as np

# Create a 4x4 matrix from 0 to 15 

matrix = np.arange(16).reshape(4, 4)

print("Original Matrix:")
print(matrix)

# Extract first row
first_row = matrix[0, :]
print("\nFirst Row:")
print(first_row)

# Extract last column
last_column = matrix[:, -1]
print("\nLast Column:")
print(last_column)

# Extract 2x2 subarray from the center
center = matrix[1:3, 1:3]
print("\nCenter 2x2 Subarray:")
print(center)
```

### explanation::
