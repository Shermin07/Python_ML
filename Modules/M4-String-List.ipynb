## ## list as queue

```python
## queue act as real life queue like we took the place

queue = []

queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(queue)

print("first item gone from queue:", queue.pop(0))


## accessing front item

print(f" now front e {queue[0]}")

print(queue.pop(0))
```

## M--4.7 List Comprehension

```python
even = []

for i in range(101):
    if i%2==0:
        even.append(i)
print(even)       
```

```python
### list comprehension::

random_list = [x  for x in range(10) if x%2==0]
print(random_list)
random_list1 = [x*2 for x in range(10) if x%2==0]
print(random_list1)
```

## M--4.8 List Comprehension with List of String

```python
## noemal list er opor loop
fruits = ["Apple","Banana","Mango"]

upper_fruits = []

for fruit in fruits:
    print(fruit)
    upper_fruits.append(fruit.upper())

print(upper_fruits) 


## make it list comprehension:

lower_fruit = [fruit.lower() for fruit in fruits]
print(lower_fruit)
```
